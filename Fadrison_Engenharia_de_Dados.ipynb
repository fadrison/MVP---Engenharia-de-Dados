função

```python
def sanitize_column_name(col_name):

    # Remove acentos
    nfkd_form = unicodedata.normalize("NFKD", col_name)

    only_ascii = nfkd_form.encode(
        "ASCII", "ignore"
    ).decode("utf-8")

    # Substitui caracteres especiais por _
    clean_name = re.sub(
        r"[^a-zA-Z0-9]",
        "_",
        only_ascii
    )

    # Remove underscores duplicados
    clean_name = re.sub(
        r"_+",
        "_",
        clean_name
    )

    # Remove underscore no início/fim
    clean_name = clean_name.strip("_")

    # Converte para minúsculo
    return clean_name.lower()
```

## **Gold (Curated)**

```python
# Views Temporárias para Execução em SQL
sdf_pagamento.createOrReplaceTempView("pagamentos")
sdf_passagem.createOrReplaceTempView("passagens")
sdf_trecho.createOrReplaceTempView("trechos")
sdf_viagem.createOrReplaceTempView("viagens")
```

```python
# Consolidação de tabela de custos Com as viagens:
df_viagens_join = spark.sql("""
SELECT
    v.identificador_do_processo_de_viagem,
    v.numero_da_proposta_pcdp,
    v.situacao,
    v.viagem_urgente,
    v.valor_outros_gastos,
    v.valor_devolucao,
    v.valor_passagens,
    v.valor_diarias,
    v.periodo_data_de_inicio,
    v.periodo_data_de_fim,

    -- Dados de Pagamento
    p.tipo_de_pagamento,
    p.valor AS valor_pagamento,
    p.nome_do_orgao_pagador,

    -- Dados de Passagem
    pass.meio_de_transporte AS transporte_passagem,
    pass.valor_da_passagem,
    pass.taxa_de_servico,

    -- Dados de Trecho
    t.numero_diarias

FROM viagens v
LEFT JOIN pagamentos p
    ON v.identificador_do_processo_de_viagem = p.identificador_do_processo_de_viagem
LEFT JOIN passagens pass
    ON v.identificador_do_processo_de_viagem = pass.identificador_do_processo_de_viagem
LEFT JOIN trechos t
    ON v.identificador_do_processo_de_viagem = t.identificador_do_processo_de_viagem
""")

# Exibir os dados
df_viagens_join.show(50)
```

```python
df_pandas = df_viagens_join.toPandas()

# 2. Salvar usando o método nativo do Pandas diretamente no caminho do Workspace
df_pandas.to_csv(
    "/Workspace/trabalho/gold/custos_viagem.csv",
    sep=";",
    index=False,
    encoding="ISO-8859-1"
)
```

```python
# identificando os 10 orgão que mais apresentam registros na base de viagens.
resultado = spark.sql("""
    SELECT
        nome_do_orgao_superior,
        COUNT(*) AS total_viagens
    FROM viagens
    GROUP BY nome_do_orgao_superior
    ORDER BY total_viagens DESC
    LIMIT 10
""")

# Exibe o resultado na tela
resultado.show(truncate=False)
```
